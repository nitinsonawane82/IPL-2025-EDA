## Loading the Libraries and Dataset

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

df=pd.read_csv('IPL_2025.csv')
df
```

## Basic information

```python
df.info()
```

## check the size of rows and columns of the dataset

```python
print(f"rows in dataset is {df.shape[0]} and columns in dataset is {df.shape[1]}")
```

## Now let's see how many columns have null values int total.

```python
df.isnull().sum()
```

## Fill null values

```python
df['second_ings_score']=df['second_ings_score'].fillna(0)
df['second_ings_wkts']=df['second_ings_wkts'].fillna(0)
df['won_by']=df['won_by'].fillna("No Result")
df['margin']=df['margin'].fillna(0)
df.isnull().sum()
```

## Now, Here comes some Basic Questions

### 1. Which team won the most matches?

```python
match_won=df['match_winner'].value_counts()
match_won
```

```python
sns.barplot(y=match_won.index,x=match_won.values,palette='viridis')
plt.title('Total Match Won')
```

### 2. Toss Decision Trends

```python
sns.countplot(x=df['toss_decision'],palette='rainbow')
plt.title("Toss Decision")
```

### 3. Toss Winner vs Match Winner

```python
count=df[df['toss_winner']==df['match_winner']]['match_id'].count()
percentage=(count*100)/df.shape[0]
print("Percentage of toss winner is match winner:",percentage.round(2))
```

### 4. How do teams win? (Runs vs Wickets)

```python
sns.countplot(x=df['won_by'],palette='viridis')
plt.title("Won By")
```

## Key Player Performances

### 1 Most "player of the match" awards

```python
count=df['player_of_the_match'].value_counts().head(10)
count
```

```python
sns.barplot(y=count.index,x=count.values,palette='viridis')
plt.title("Top 10 Players")
```

### 2 top Scorers

```python
high_scorers=df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high_scorers
```

```python
high_scorers.plot(kind='barh')
```

### 10 best bowling figures

```python
df['highest_wickets']=df['best_bowling_figure'].apply(lambda x:x.split('--')[0])   #create new column "highest_wickets"
df['highest_wickets']=df['highest_wickets'].astype(int)
bowler=df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
bowler
```

```python
bowler.plot(kind='barh')
```

## Venue Analysis

### Most Matches Played by Venue

```python
venue=df['venue'].value_counts().head(10)
venue
```

```python
sns.barplot(y=venue.index,x=venue.values,palette='viridis')
```
